```python
import pandas as pd

df_students = pd.DataFrame({
    'StudentID': [56, 2, 3, 4, 5],
    'FirstName': ['Krrish', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'LastName': ['P', 'Johnson', 'Brown', 'Davis', 'Miller'],
    'Major': ['AI', 'Physics', 'Mathematics', 'Chemistry', 'Biology'],
    'EntryYear': [2025, 2020, 2022, 2021, 2023],
    'GPA': [9.2, 7.9, 9.1, 7.3, 8.8]
})

print("Initial df_students:")
display(df_students)

students1 = pd.DataFrame({
    'StudentID': [51, 52, 55],
    'FirstName': ['Uma', 'Dhanush', 'Rohit'],
    'LastName': ['Rani', 'Iyer', 'Kumar'],
    'Major': ['Physics', 'Art', 'Math'],
    'EntryYear': [2024, 2023, 2025],
    'GPA': [7.1, 8.5, 8.4]
})

#created randoom datasets

students2 = pd.DataFrame({
    'StudentID': [53, 54, 56],
    'FirstName': ['Walter', 'Chahal', 'Krrish'],
    'LastName': ['White', 'Singh', 'P'],
    'Major': ['Chemistry', 'History', 'AI'],
    'EntryYear': [2022, 2024, 2025],
    'GPA': [9.2, 7.9, 9.2]
})

# Concatenate the new student dataframes (part 1 and part 2) together
all_new_students = pd.concat([students1, students2], ignore_index = True)
#assigns new indices to elements

print("Concatenated new student records:")
display(all_new_students)

# Using an 'outer' merge on 'StudentID' to include all students from both DataFrames
updated_df1 = pd.merge(df_students, all_new_students, on='StudentID', how='outer')
updated_df2 = pd.merge(df_students, all_new_students, on='FirstName', how='inner') #using inner to merge dataframes according to First Name

# Display the updated df_students DataFrame

print("\nOriginal df_students head:")
display(df_students.head())

print("\nUpdated df_students (after merging new students) tail:")
display(updated_df1.tail())
display(updated_df2.tail())
```

```python
import pandas as pd
import numpy as np

data_example = {'CityA': [10, 20], 'CityB': [15, 25]}
df_example = pd.DataFrame(data_example, index=['Temp', 'Humidity'])
print("Original DataFrame (df_example):")
display(df_example)

stacked_series = df_example.stack()
print("\nStacked Series:")
display(stacked_series)

unstacked_df = stacked_series.unstack()
print("\nUnstacked DataFrame:")
display(unstacked_df)


# Create two simple Series
series_one = pd.Series([100, 200], index=['Item1', 'Item2'])
series_two = pd.Series([300, 400], index=['Item3', 'Item4'])

print("Series One:")
display(series_one)
print("\nSeries Two:")
display(series_two)

# 3. pd.concat() with keysCreate a hierarchical index
concatenated_with_keys = pd.concat([series_one, series_two], keys=['Dataset_X', 'Dataset_Y'])
print("\nConcatenated Series with hierarchical index:\n")
display(concatenated_with_keys)

unstacked = concatenated_with_keys.unstack()
print("Unstacked Concatenated Series (showing how keys would become columns):")
display(unstacked)
```
