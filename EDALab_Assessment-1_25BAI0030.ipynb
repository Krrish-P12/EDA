```python
import pandas as pd
import numpy as np
from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv("/content/drive/MyDrive/chess_games.csv",encoding = 'latin1')

#Displays the last 5 rows of the dataframe
display(df.tail())

#Displays the first 6 rows of the dataframe
display(df.head(6))

print("Info and description of the dataframe\n");
df.info() #gives the data type of the rows and column entities
df.describe() #gives various details like central tendencies etc

winner_stats = df['winner'].value_counts()
print(winner_stats)

print("\nLong games count\n")
long_games = df[df['turns'] > 100]
display(long_games.head())

#prints the rating difference
df['rating_difference'] = abs(df['white_rating'] - df['black_rating'])
display(df.head())

# Average turns by winner
avgturnswinner = df.groupby('winner')['turns'].mean()
print(avgturnswinner)

# Sorting the DataFrame by turns in descending order
print(" Sorted Data: Turns in Descending Order\n")
dfsorted = df.sort_values(by='turns', ascending=False)
display(dfsorted.head())

#New column total_rating by summing white and black ratings
print("Data with New Column: Total Rating\n")
df['total_rating'] = df['white_rating'] + df['black_rating']
display(df.head())

# Counting the number of games for each opening using value counts
print("Value Counts: Top 5 Opening Full Names\n")
opening_counts = df['opening_fullname'].value_counts()
print(opening_counts.head())

#Showing top 6 games with Queen's Gambit
print("Games with Caro-Kann Opening\n")
QG = df[df['opening_fullname'] == "Queen's Gambit"]
display(QG.head(6))
```
