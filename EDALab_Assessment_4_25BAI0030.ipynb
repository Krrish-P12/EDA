```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer

#replacing and displaying all null values using different imputation techniques
df_new = pd.DataFrame({
    'Experience_Years': [2, 5, np.nan, 10, 3, 7, np.nan, 1],
    'Monthly_Income': [3000, 5500, 4000, np.nan, 3200, 6000, 4500, np.nan],
    'Employment_Status': ['Full-time', 'Part-time', 'Full-time', 'Contract', np.nan, 'Full-time', 'Part-time', 'Full-time'],
    'Project_Count': [1, np.nan, 3, 5, 2, np.nan, 4, 1]
})

print("Original DataFrame with missing values:")
display(df_new)

# 1. Mean Imputation
mean_imputer_exp = SimpleImputer(strategy='mean')
df_new['Experience_Years_Mean'] = mean_imputer_exp.fit_transform(df_new[['Experience_Years']])

# 2. Median Imputation
median_imputer_income = SimpleImputer(strategy='median')
df_new['Monthly_Income_Median'] = median_imputer_income.fit_transform(df_new[['Monthly_Income']])

# 3. Mode Imputation
mode_imputer_status = SimpleImputer(strategy='most_frequent')
df_new['Employment_Status_Mode'] = mode_imputer_status.fit_transform(df_new[['Employment_Status']])[:, 0]

# 4. MICE Imputation
mice_cols = ['Experience_Years', 'Project_Count']
mice_imputer_data = IterativeImputer(max_iter=5, random_state=0)
df_mice_subset = df_new[mice_cols].copy()
df_mice_subset_imputed = mice_imputer_data.fit_transform(df_mice_subset)

df_new['Experience_Years_MICE'] = df_mice_subset_imputed[:, 0]
df_new['Project_Count_MICE'] = df_mice_subset_imputed[:, 1]

display(df_new)
```
