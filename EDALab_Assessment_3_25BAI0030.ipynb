```python
import numpy as np
import pandas as pd
data = np.arange(10, 25).reshape(5, 3)
dfx = pd.DataFrame(data, index=['apple', 'banana', 'kiwi',
'grapes', 'mango'], columns=['shop_a', 'shop_b', 'shop_c'])

#Add missing value
dfx['shop_d'] = np.nan
dfx.loc['orange'] = np.arange(100, 104)
dfx.loc['pear'] = np.nan
dfx['shop_e'] = np.nan
dfx['shop_d']['banana'] = 50.0
dfx.loc['kiwi', 'shop_c'] = np.nan
display(dfx)

print(dfx.isnull())
x = dfx.isnull().sum()

print("Number of missing values in each column: \n", x)
print("Total number of missing values: ",x.sum())

#empty rows can be dropped by dropna()

print(dfx.dropna()) #empty dataset is returned since all rows have atleast one NaN value

#To avoid this dropna(how = "all") and axis = 1 for dropping columns with all NaN values instead of rows

print(dfx.dropna(how = "all", axis = 1), "\n") #this drops the last column which has all NaN values

print("Dropping columns with minimum 3 NaN values:\n ", dfx.dropna(thresh = 3, axis = 1), "\n")

newdf = dfx.fillna('Mango') #replace NaN values with 'Mango'
print(newdf)
```
